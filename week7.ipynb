```python
# ShowFib.py
"""
Investigates the Fibonacci sequence using while-loops and
explores its relation to the golden ratio:

    golden ratio r = (1 + sqrt(5)) / 2

Given two consecutive Fibonacci numbers:
    next = current + predecessor

Indexing convention:
    Fib(0) = 0
    Fib(1) = 1
    Fib(2) = 1
    Fib(3) = 2
    ...

This script solves four problems:
1. Print Fibonacci numbers 1 through 15 and their consecutive ratios
2. Print Fibonacci numbers until the ratio approximates the golden ratio
3. Find the smallest Fibonacci number > 1,000,000
4. Find the largest Fibonacci number < 1,000,000
"""

import math

# ---------------------------------------------
# Problem 1: Print Fibonacci numbers 1 through 15
# ---------------------------------------------
k = 1          # Fibonacci index
x = 0          # Fib(k-1)
y = 1          # Fib(k)
z = 1          # Fib(k+1)
ratio = z / y  # ratio of next/current

print('\n  k     kth Fib      kth Ratio ')
print('--------------------------------------')

while k <= 20:
    print(f'{k:3d} {y:10d}     {ratio:14.12f}')
    k += 1
    x = y       # update Fib(k-1)
    y = z       # update Fib(k)
    z = x + y   # compute next Fibonacci number
    ratio = z / y

# ---------------------------------------------
# Problem 2: Print Fibonacci numbers until ratio approximates golden ratio
# ---------------------------------------------
r = (1 + math.sqrt(5)) / 2  # golden ratio
k = 1
x = 0
y = 1
z = 1
error = abs(z / y - r)

print('\n  k     kth Fib    kth Ratio Error ')
print('--------------------------------------')

while error >= 1e-9:  # continue until ratio is very close to golden ratio
    print(f'{k:3d} {y:10d}     {error:14.12f}')
    k += 1
    x = y
    y = z
    z = x + y
    error = abs(z / y - r)

# Print the final Fibonacci number where error < threshold
print(f'{k:3d} {y:10d}     {error:14.12f}')

```

```python
# ---------------------------------------------
# Problem: Smallest Fibonacci number > 1,000,000
# ---------------------------------------------
k = 1
x = 0
y = 1
z = 1

while y < 1_000_000:
    k += 1
    x = y
    y = z
    z = x + y

print(f'\nThe smallest Fibonacci number > 1,000,000 is Fib {k} = {y}')

# ---------------------------------------------
# Problem 4: Largest Fibonacci number < 1,000,000
# ---------------------------------------------
k = 1
x = 0
y = 1
z = 1

while z < 1_000_000:
    k += 1
    x = y
    y = z
    z = x + y

print(f'\nThe largest Fibonacci number < 1,000,000 is Fib {k} = {y}')
```

```python
def inS(a, b, x, y):
    """Returns True if (x,y) is inside the square with vertices
    (a,b), (a+1,b), (a,b+1), and (a+1,b+1).
    Otherwise, returns False."""

    xOK = (a <= x <= a + 1)
    yOK = (b <= y <= b + 1)
    return (xOK and yOK)


a1 =2
a2=6.2
b1=3
b2=3.2
x=2.5
y=3.5
z_union = inS(a1, b1, x, y) or inS(a2, b2, x, y)
z_intersection = inS(a1, b1, x, y) and inS(a2, b2, x, y)
print(z_union, z_intersection)
```

## assert

In Python, assert is a keyword used as a debugging aid to test if a specific condition is true. If the condition passes (True), the program continues executing normally. If the condition fails (False), Python immediately halts the program and raises an AssertionError

```python
import math

def sqrt(x):
    assert isinstance(x, (int, float)), \
        "x must be type int or float"
    assert x >= 0, "Dr. Wang said x must be nonnegative"
    return math.sqrt(x)

print(sqrt("s"))
```

```python
# ShowAssert.py
"""Demonstrates how assert statements can be used to check multiple preconditions."""

import string


def TimeToMinutes(T):
    """
    Returns the total number of minutes that have elapsed from midnight
    to the time specified by T.

    Precondition:
        T is a string in the format 'hh:mm'
        where hh is between 00 and 23
        and mm is between 00 and 59.

    Valid examples: '00:00', '12:34', '23:59'
    Invalid examples: '1:23', '24:01', '01:78', 'xy:zw'
    """

    # Validate preconditions
    assert isinstance(T, str), "T must be a string."
    assert len(T) == 5, "Time string must have length 5."
    assert T[2] == ":", "The third character must be a colon."
    assert T[0] in string.digits and T[1] in string.digits, \
        "The hour portion must be numeric."
    assert T[3] in string.digits and T[4] in string.digits, \
        "The minute portion must be numeric."

    hours = int(T[0:2])
    minutes = int(T[3:5])

    assert 0 <= hours <= 23, "Hour must satisfy 0 <= hh <= 23."
    assert 0 <= minutes <= 59, "Minute must satisfy 0 <= mm <= 59."

    total_minutes = 60 * hours + minutes

    # Postcondition check
    assert isinstance(total_minutes, int), "Return value must be an integer."

    return total_minutes


def MinutesToTime(M):
    """
    Returns a time string in the format 'hh:mm' representing
    M minutes past midnight.

    Precondition:
        M is an integer satisfying 0 <= M < 24*60.
    """

    # Validate preconditions
    assert isinstance(M, int), "M must be an integer."
    assert 0 <= M < 24 * 60, "M must satisfy 0 <= M < 1440."

    # Compute hours and minutes
    hours = M // 60
    minutes = M % 60

    # Format with leading zeros
    T = f"{hours:02d}:{minutes:02d}"

    # Postcondition validation
    assert isinstance(T, str), "Return value must be a string."
    assert len(T) == 5, "Time string must have length 5."
    assert T[2] == ":", "The third character must be a colon."
    assert T[0] in string.digits and T[1] in string.digits, \
        "Hour portion must be numeric."
    assert T[3] in string.digits and T[4] in string.digits, \
        "Minute portion must be numeric."

    return T


# Application script
if __name__ == "__main__":
    """
    Prompts the user for a time string, converts it to minutes,
    and then converts the minutes back to a time string.
    """

    T = input("Enter a time string in the form hh:mm: ")
    print("Input time:", T)

    M = TimeToMinutes(T)
    print("Total minutes:", M)

    T = MinutesToTime(M)
    print("Converted back to time:", T)
```

## try-except

In Python, try and except blocks are used for exception handling, which prevents your program from crashing when a runtime error occurs. Instead of halting execution with a messy error message, it allows your code to fail gracefully, log the problem, or take an alternative action.

How It Works
1. try block: You place the "risky" code that might cause an error inside this block.
2. except block: Python monitors the try block. If an error occurs, it instantly jumps out of that block and runs the code inside the except block.

```python
"""
Illustrates the use of try-except.
Encourages the user to input a nonnegative integer because
that is what the factorial function requires
"""

import math

while True:
    N = input("Enter a nonnegative integer: ")

    try:
        # Convert the input string to an int.
        N = int(N)

        if N < 0:
            print("N must be nonnegative.")
            continue

        # Valid input. Terminate the loop.
        break

    except ValueError:
        # Invalid input. Inform the user and the iteration continues.
        print("N must have type int")

M = math.factorial(N)

print(f"\n\n{N}! = {M}")
```

```python
"""
SqrtAssert.py
Shows how the assert statement can be used to check
that the preconditions of a function are satisfied
and that the returned value satisfies the specification.
"""

def sqrt(x):
    """
    Returns an approximate square root L of x such that
    |L*L - x| <= 0.001

    Precondition:
        x is a positive number.
    """

    assert x > 0, "The sqrt function requires a positive argument."

    L = float(x)

    # Newton's method iteration (4 iterations as in original code)
    L = (L + x / L) / 2
    L = (L + x / L) / 2
    L = (L + x / L) / 2


    assert abs(L * L - x) <= 0.001, "Inaccurate Square Root"

    return L


if __name__ == "__main__":
    x = float(input("Enter a positive number: "))
    z = sqrt(x)

    print(f"sqrt({x:5.3f}) = {z:10.6f}")
```
