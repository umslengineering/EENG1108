### 1. Python f-strings

```python
r = 18.1234567
print("the radius is ", r)
print("the radius is %d" %r)
print("the radius is %f" %r)
print("the radius is %10.2f" %r)
print("------------------------")
print(f"the radius is {r}")
print(f"the radius is {r:.2f}")
print("------------------------")
word = "haveyoueatenyet"
print(f"{word:>30}") #right-aligned
print(f"{word:<30}") #left-aligned
```

### 2. Surface Area Increase Calculations of a Sphere

```python
# surface area of a sphere is 4*pi*r**2
# increase the radius, causing the increase of the surface area
from math import pi

r = float(input("enter the radius in km:"))
inc = float(input("Enter the increament of radius in meters:"))

# increament in km
dr = inc/1e3

# solve the increase of area
# method 1
dS_1 = 4*pi*((r+dr)**2 - r**2)
print(f"method 1 area increase : {dS_1} km square")

# method 2
dS_2 = 4*pi*(dr**2 + 2*dr*r)
print(f"method 2 area increase : {dS_2} km square")

# method 3
dS_3 = 4*pi*( 2*dr*r)
print(f"method 3 area increase : {dS_3} km square")
```
