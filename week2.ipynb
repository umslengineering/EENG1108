```python
title = "SpaceExplorer" #string
launches = 3  # int
fuel_percentage = 87.5  # float
is_ready = True #boolean
crew = ["Ari", "Bo", "BB8"] #list
ship = {"name":"SpaceExplorer", "speed":"9000"} # dictionary

for values in [title, launches, fuel_percentage, is_ready, crew, ship]:
  print(repr(values) , "->", type(values).__name__)
```
