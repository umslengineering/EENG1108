```python
# energy boost for spaceship

def calculate_energy(energy_percentage, boost=0):
  return min(100, energy_percentage+boost)

print(calculate_energy(93))
```

```python
# calculate the radius from the area
from math import sqrt,pi

A = float(input("Enter the area "))

def calc_r(A):
  return sqrt(A/pi)

print("the radius is:", calc_r(A))
```
