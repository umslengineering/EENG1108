# 1. Tuples

A **tuple** is an **ordered, immutable** sequence of items. Items can be of any type, and duplicates are allowed.

- Ordered: items keep their position, accessed by index
- Immutable: you cannot add, remove, or change items after creation
- Written with parentheses `()`, but it is the **comma** that makes a tuple

## 1.1 Creating tuples

```python
empty = ()
nums = (1, 2, 3)
mixed = (1, "hello", 3.14, True)
no_parens = 10, 20, 30          # parentheses are optional
from_list = tuple([1, 2, 3])    # convert from any iterable
from_str = tuple("abc")

print(empty, nums, mixed, no_parens, from_list, from_str, sep="\n")
```

### Single-element tuple gotcha
A trailing comma is required. `(5)` is just the integer 5 in parentheses.

```python
not_a_tuple = (5)
a_tuple = (5,)
print(type(not_a_tuple), type(a_tuple))
```

## 1.2 Indexing and slicing

```python
t = ("a", "b", "c", "d", "e")

print(t[0])        # first
print(t[-1])       # last
print(t[1:4])      # index 1 up to (not including) 4
print(t[::2])      # every 2nd item
print(t[::-1])     # reversed copy
```

## 1.3 Immutability

```python
t = (1, 2, 3)
try:
    t[0] = 99
except TypeError as e:
    print("TypeError:", e)
```

**Caveat:** a tuple is immutable, but if it *contains* a mutable object (like a list), that inner object can still change.

```python
t = (1, [2, 3], 4)
t[1].append(99)
print(t)            # (1, [2, 3, 99], 4) -- the tuple still holds the same list object
```

## 1.4 Tuple operations and methods

```python
a = (1, 2, 3)
b = (4, 5)

print(a + b)            # concatenation -> new tuple
print(a * 2)            # repetition
print(2 in a)           # membership
print(len(a), min(a), max(a), sum(a))

t = (1, 2, 2, 3, 2)
print(t.count(2))       # how many times 2 appears
print(t.index(3))       # index of first 3
```

Tuples have only two methods, `count()` and `index()`, precisely because they cannot be modified.

## 1.5 Packing and unpacking

```python
# Packing
point = (3, 4)

# Unpacking
x, y = point
print(x, y)

# Swap without a temp variable
a, b = 1, 2
a, b = b, a
print(a, b)

# Extended unpacking with *
first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)      # middle is a list

# Ignore values with _
name, _, age = ("Alice", "ignored", 30)
print(name, age)
```

Unpacking requires the number of variables to match (unless `*` is used), otherwise you get a `ValueError`.

## 1.6 Returning multiple values from a function

```python
def min_max(values):
    return min(values), max(values)     # returns a tuple

lo, hi = min_max([4, 1, 9, 3])
print(lo, hi)
```

## 1.7 Iterating over tuples

```python
colors = ("red", "green", "blue")

for c in colors:
    print(c)

for i, c in enumerate(colors):
    print(i, c)

for a, b in zip(colors, (1, 2, 3)):
    print(a, b)
```

## 1.8 Nested tuples

```python
matrix = ((1, 2, 3), (4, 5, 6))
print(matrix[1][2])         # 6

for row in matrix:
    for val in row:
        print(val, end=" ")
```

## 1.9 Tuples are hashable (if all items are hashable)

Because they are immutable, tuples can be **dictionary keys** and **set members**; lists cannot.

```python
# （Longitude， latitude）
locations = {(40.7, -74.0): "New York", (51.5, -0.1): "London"}
print(locations[(40.7, -74.0)])

try:
    bad = {[1, 2]: "list key"}
except TypeError as e:
    print("TypeError:", e)
```

## 1.10 Named tuples

`collections.namedtuple` (or `typing.NamedTuple`) gives tuple fields readable names while keeping immutability and memory efficiency.

```python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)

print(p)
print(p.x, p.y)         # by name
print(p[0], p[1])       # still works by index
print(p._asdict())      # convert to dict
print(p._replace(x=10)) # returns a new tuple
```

```python
from typing import NamedTuple

class Employee(NamedTuple):
    name: str
    age: int
    dept: str = "Eng"

e = Employee("Sam", 28)
print(e)
```

## 1.11 Tuple vs list

| Feature | Tuple | List |
|---|---|---|
| Syntax | `(1, 2)` | `[1, 2]` |
| Mutable | No | Yes |
| Hashable (as dict key) | Yes, if contents are hashable | No |
| Methods | `count`, `index` | many (`append`, `sort`, ...) |
| Typical use | Fixed records, multiple return values, keys | Collections that change |
| Speed / memory | Slightly faster, smaller | Slightly slower, larger |

```python
import sys
print(sys.getsizeof((1, 2, 3)), sys.getsizeof([1, 2, 3]))
```

### Tuple summary
- Use tuples for **fixed collections** of related values (coordinates, RGB, DB rows).
- Single-item tuple needs a trailing comma: `(1,)`.
- Unpacking is one of the most useful tuple features.
- Immutability is shallow: contained mutable objects can still change.

# 2. Dictionaries

A **dictionary** (`dict`) stores **key-value pairs**. It is **mutable**, provides fast lookup by key, and (since Python 3.7) **preserves insertion order**.

- Keys must be **unique** and **hashable** (strings, numbers, tuples of hashables, ...)
- Values can be anything
- Written with curly braces `{key: value}`

## 2.1 Creating dictionaries

```python
empty = {}
person = {"name": "Alice", "age": 30, "city": "Paris"}

d1 = dict(name="Bob", age=25)                   # keyword args (keys must be valid identifiers)
d2 = dict([("a", 1), ("b", 2)])                 # from list of pairs
d3 = dict(zip(["x", "y", "z"], [10, 20, 30]))   # from two sequences
d4 = dict.fromkeys(["a", "b", "c"], 0)          # same value for every key

print(person, d1, d2, d3, d4, sep="\n")
```

## 2.2 Accessing values

```python
person = {"name": "Alice", "age": 30}

print(person["name"])               # KeyError if missing
print(person.get("city"))           # None if missing
print(person.get("city", "St. Louis"))    # custom default

try:
    person["city"]
except KeyError as e:
    print("KeyError:", e)
```

## 2.3 Adding, updating, removing

```python
d = {"a": 1, "b": 2}

d["c"] = 3                  # add
d["a"] = 100                # update
d.update({"b": 20, "d": 4}) # update many (add or overwrite)
print(d)

print(d.pop("b"))           # remove key, return value
print(d.pop("zzz", None))   # safe pop with default
del d["c"]                  # remove key
print(d.popitem())          # remove and return last inserted pair
print(d)

d.clear()
print(d)
```

### `setdefault`
Returns the value for a key, inserting the default first if the key is missing.

```python
d = {}
d.setdefault("fruits", []).append("apple")
d.setdefault("fruits", []).append("banana")
print(d)
```

## 2.4 Membership, length

```python
d = {"a": 1, "b": 2}
print("a" in d)           # checks KEYS
print(1 in d)             # False: 1 is a value, not a key
print(1 in d.values())    # True
print(len(d))
```

## 2.5 Views: keys, values, items

```python
d = {"a": 1, "b": 2, "c": 3}

print(d.keys())
print(d.values())
print(d.items())

# Views are live: they reflect later changes
keys = d.keys()
d["z"] = 26
print(keys)

print(list(d.items()))
```

## 2.6 Iterating

```python
scores = {"Ann": 90, "Ben": 75, "Cal": 82}

for k in scores:                    # keys by default
    print(k)

for v in scores.values():
    print(v)

for k, v in scores.items():         # most common pattern
    print(f"{k}: {v}")
```

> **Warning:** don't add or remove keys while iterating over a dict (it raises `RuntimeError`). Iterate over `list(d)` or build a new dict instead.

```python
d = {"a": 1, "b": 2, "c": 3}
for k in list(d):               # snapshot of keys
    if d[k] % 2 == 1:
        del d[k]
print(d)
```

## 2.7 Dictionary comprehensions

```python
squares = {n: n**2 for n in range(1, 6)}
print(squares)

evens = {k: v for k, v in squares.items() if v % 2 == 0}
print(evens)

# Invert a dict (values must be unique and hashable)
inv = {v: k for k, v in squares.items()}
print(inv)
```

## 2.8 Merging dictionaries

```python
a = {"x": 1, "y": 2}
b = {"y": 20, "z": 30}

print({**a, **b})       # unpacking (works in 3.5+); later wins
print(a | b)            # merge operator (3.9+); later wins

a |= b                  # in-place merge (3.9+)
print(a)
```

## 2.9 Nested dictionaries

```python
students = {
    "alice": {"age": 20, "grades": [90, 85]},
    "bob":   {"age": 22, "grades": [70, 88]},
}

print(students["alice"]["grades"][0])

students["carol"] = {"age": 21, "grades": []}
students["bob"]["age"] += 1

for name, info in students.items():
    print(name, info["age"], info["grades"])

# Safe deep access
print(students.get("dave", {}).get("age"))
```

## 2.10 Copying: shallow vs deep

```python
import copy

original = {"a": [1, 2], "b": 2}

alias = original                  # same object
shallow = original.copy()         # new dict, SAME inner objects
deep = copy.deepcopy(original)    # fully independent

original["a"].append(3)
print(alias["a"], shallow["a"], deep["a"])
```

## 2.11 Sorting dictionaries

```python
scores = {"Ann": 90, "Ben": 75, "Cal": 82}

print(sorted(scores))                                   # keys sorted
print(sorted(scores.items()))                           # by key
print(sorted(scores.items(), key=lambda kv: kv[1]))     # by value
print(dict(sorted(scores.items(), key=lambda kv: kv[1], reverse=True)))

print(max(scores, key=scores.get))                      # key with the max value
```

## 2.12 Useful patterns

### Counting

```python
text = "mississippi"

counts = {}
for ch in text:
    counts[ch] = counts.get(ch, 0) + 1
print(counts)

# Better: Counter
from collections import Counter
c = Counter(text)
print(c)
print(c.most_common(2))
```

### Grouping with `defaultdict`

```python
from collections import defaultdict

fruits = ["apple", "banana", "apple", "cherry", "banana", "apple"]

#defaultdict(int) creates a dictionary
#that automatically gives a new key the default value 0.
counts = defaultdict(int)  # missing key -> 0
# means: “Create a dictionary for counting.
# If I use a key that isn’t in it yet, start that key at zero.”

for fruit in fruits:
    counts[fruit] += 1

print(dict(counts))
```

### Dispatch table (functions as values)

```python
def add(a, b): return a + b
def sub(a, b): return a - b

ops = {"+": add, "-": sub}
print(ops["+"](5, 3), ops["-"](5, 3))
```

### Tuples as dictionary keys

```python
grid = {(0, 0): "start", (2, 3): "goal"}
print(grid.get((2, 3)))
```

## 2.13 Key requirements

```python
d = {}
d[1] = "int"
d["a"] = "str"
d[(1, 2)] = "tuple"
d[3.0] = "float"
d[True] = "bool"        # True == 1 and hash(True) == hash(1) -> overwrites key 1!

print(d)

try:
    d[[1, 2]] = "list"
except TypeError as e:
    print("TypeError:", e)
```

## 2.14 Time complexity (average case)

| Operation | Complexity |
|---|---|
| `d[key]`, `d.get(key)` | O(1) |
| `d[key] = value` | O(1) |
| `del d[key]`, `d.pop(key)` | O(1) |
| `key in d` | O(1) |
| Iterate all items | O(n) |
| `value in d.values()` | O(n) |
| `d.copy()` | O(n) |

## 2.15 Method quick reference

| Method | Purpose |
|---|---|
| `d.get(k, default)` | Safe lookup |
| `d.setdefault(k, default)` | Get, inserting default if missing |
| `d.update(other)` | Merge/overwrite from another dict or pairs |
| `d.pop(k, default)` | Remove key, return its value |
| `d.popitem()` | Remove and return last inserted pair |
| `d.keys()`, `d.values()`, `d.items()` | Live views |
| `d.copy()` | Shallow copy |
| `d.clear()` | Remove everything |
| `dict.fromkeys(keys, v)` | New dict with same value for each key |

### Dictionary summary
- Fast key-based lookup; keys unique and hashable.
- Prefer `.get()` / `setdefault` / `defaultdict` over manual key-existence checks.
- Use `.items()` for loops, comprehensions for transformations, `|` or `{**a, **b}` for merging.
- `.copy()` is shallow; use `copy.deepcopy` for nested data.

# 3. Tuples vs Dictionaries

| | Tuple | Dictionary |
|---|---|---|
| Structure | Ordered sequence | Key-value mapping |
| Access | By integer index | By key |
| Mutable | No | Yes |
| Duplicates | Items may repeat | Keys unique; values may repeat |
| Use when | Fixed group of values, positional meaning | Labeled data, fast lookup |

```python
# Same record, two representations
record_tuple = ("Alice", 30, "Paris")
record_dict = {"name": "Alice", "age": 30, "city": "Paris"}

print(record_tuple[1])          # you must remember what index 1 means
print(record_dict["age"])       # self-documenting
```
